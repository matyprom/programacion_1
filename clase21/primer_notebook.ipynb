```python
import json

def convertir_txt_a_json(archivo_txt, archivo_json):
    datos = []

    with open(archivo_txt, "r", encoding="utf-8") as archivo:
        for linea in archivo:
            linea = linea.strip()
            if not linea:
                continue

            if ":" in linea:
                clave, valor = linea.split(":", 1)
                datos.append({
                    "clave": clave.strip(),
                    "valor": valor.strip()
                })
            else:
                datos.append({"valor": linea})

    with open(archivo_json, "w", encoding="utf-8") as archivo:
        json.dump(datos, archivo, ensure_ascii=False, indent=4)

# Ejemplo de uso
convertir_txt_a_json("datos.txt", "datos.json")
```
